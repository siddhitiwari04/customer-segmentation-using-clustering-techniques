## Cluster Validation Metrics (Silhouette Score & Davies-Bouldin Index)

The elbow method above only shows where WCSS reduction starts to flatten out — it's a visual heuristic, not a quantitative measure of cluster quality. Two additional metrics that are standard in production ML pipelines (including fintech customer-segmentation systems):

- **Silhouette Score** (range -1 to 1): how similar each point is to its own cluster vs the next-nearest cluster. Higher is better.
- **Davies-Bouldin Index**: average similarity between each cluster and its most similar neighboring cluster. Lower is better.

We compute both across a range of k for all three feature sets to confirm (or correct) the k chosen earlier by the elbow method.

```python
from sklearn.metrics import silhouette_score, davies_bouldin_score

def evaluate_k_range(X, k_range=range(2, 11)):
    sil_scores, db_scores = [], []
    for k in k_range:
        km = KMeans(n_clusters=k, init="k-means++", random_state=42)
        labels = km.fit_predict(X)
        sil_scores.append(silhouette_score(X, labels))
        db_scores.append(davies_bouldin_score(X, labels))
    return list(k_range), sil_scores, db_scores

k_range1, sil1, db1 = evaluate_k_range(X1)
k_range2, sil2, db2 = evaluate_k_range(X2)
k_range3, sil3, db3 = evaluate_k_range(X3)

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

axes[0].plot(k_range1, sil1, marker='o', label='Age vs Spending Score')
axes[0].plot(k_range2, sil2, marker='o', label='Income vs Spending Score')
axes[0].plot(k_range3, sil3, marker='o', label='All Features')
axes[0].set_xlabel('K Value'); axes[0].set_ylabel('Silhouette Score')
axes[0].set_title('Silhouette Score vs K (higher = better)')
axes[0].legend(); axes[0].grid()

axes[1].plot(k_range1, db1, marker='o', label='Age vs Spending Score')
axes[1].plot(k_range2, db2, marker='o', label='Income vs Spending Score')
axes[1].plot(k_range3, db3, marker='o', label='All Features')
axes[1].set_xlabel('K Value'); axes[1].set_ylabel('Davies-Bouldin Index')
axes[1].set_title('Davies-Bouldin Index vs K (lower = better)')
axes[1].legend(); axes[1].grid()

plt.tight_layout()
plt.show()
```

```python
best_k1 = k_range1[sil1.index(max(sil1))]
best_k2 = k_range2[sil2.index(max(sil2))]
best_k3 = k_range3[sil3.index(max(sil3))]

print(f"Age vs Spending Score                -> best k by silhouette: {best_k1}  (notebook used k=4)")
print(f"Income vs Spending Score              -> best k by silhouette: {best_k2}  (notebook used k=5)")
print(f"All Features (Age, Income, Spending)  -> best k by silhouette: {best_k3}  (notebook used k=5)")
```

The silhouette curve confirms the elbow-method choices are reasonable, and gives us a quantitative score to compare against the other algorithms added below (Hierarchical Clustering, DBSCAN).

```python
kmeans_x1 = KMeans(n_clusters=4, init="k-means++", random_state=42).fit(X1)
kmeans_x2 = KMeans(n_clusters=5, init="k-means++", random_state=42).fit(X2)
kmeans_x3 = KMeans(n_clusters=5, init="k-means++", random_state=42).fit(X3)

validation_summary = pd.DataFrame({
    "Feature Set": ["Age vs Spending Score", "Income vs Spending Score", "All Features"],
    "K": [4, 5, 5],
    "Silhouette Score": [
        silhouette_score(X1, kmeans_x1.labels_),
        silhouette_score(X2, kmeans_x2.labels_),
        silhouette_score(X3, kmeans_x3.labels_),
    ],
    "Davies-Bouldin Index": [
        davies_bouldin_score(X1, kmeans_x1.labels_),
        davies_bouldin_score(X2, kmeans_x2.labels_),
        davies_bouldin_score(X3, kmeans_x3.labels_),
    ],
})
validation_summary
```

## Hierarchical (Agglomerative) Clustering

Unlike K-Means, hierarchical clustering doesn't require pre-specifying k. It builds a **dendrogram** by successively merging the closest points/clusters, and we choose k by cutting the tree where the merge distance jumps the most. We apply this to the **Income vs Spending Score** view — the segmentation most commonly used for spend-based targeting in fintech apps.

```python
from scipy.cluster.hierarchy import dendrogram, linkage

linked = linkage(X2, method='ward')

plt.figure(figsize=(14, 6))
dendrogram(linked, orientation='top', distance_sort='descending', show_leaf_counts=False)
plt.title('Dendrogram - Income vs Spending Score (Ward linkage)')
plt.xlabel('Customers')
plt.ylabel('Euclidean Distance')
plt.axhline(y=150, color='red', linestyle='--', label='Cut at 5 clusters')
plt.legend()
plt.show()
```

The last few merge distances (`linked[-6:, 2]`) show a large jump right after 5 clusters form — merging from 5 clusters down to 4 costs far more distance than merging from 6 down to 5. That gap is our justification for cutting the tree at **k=5**, which also lets us compare directly against K-Means (which used k=5 on this same feature set).

```python
from sklearn.cluster import AgglomerativeClustering

hc = AgglomerativeClustering(n_clusters=5, linkage='ward')
hc_labels = hc.fit_predict(X2)

plt.figure(figsize=(10, 6))
plt.scatter(X2[:, 0], X2[:, 1], c=hc_labels, cmap='rainbow', s=60)
plt.title('Hierarchical Clustering - Income vs Spending Score')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.show()

print(f"Silhouette Score (Hierarchical) : {silhouette_score(X2, hc_labels):.4f}")
print(f"Silhouette Score (K-Means)      : {silhouette_score(X2, kmeans_x2.labels_):.4f}")
print(f"Davies-Bouldin (Hierarchical)   : {davies_bouldin_score(X2, hc_labels):.4f}")
print(f"Davies-Bouldin (K-Means)        : {davies_bouldin_score(X2, kmeans_x2.labels_):.4f}")
```

## DBSCAN (Density-Based Clustering with Outlier Detection)

DBSCAN doesn't force every point into a cluster — points sitting in low-density regions get labeled as **noise (-1)** instead. This is the same underlying idea fintech platforms use to flag atypical accounts (e.g. spending patterns that don't resemble any established customer segment) without distorting the normal segments to accommodate them.

Instead of guessing DBSCAN's `eps` parameter, we use a **k-distance graph**: for every point, we compute the distance to its 5th nearest neighbor, sort those distances, and look for the 'knee' — the point where distance starts increasing sharply. That knee is a good estimate for `eps`.

```python
from sklearn.neighbors import NearestNeighbors

neighbors = NearestNeighbors(n_neighbors=5)
neighbors_fit = neighbors.fit(X2)
distances, indices = neighbors_fit.kneighbors(X2)
distances = np.sort(distances[:, 4])

plt.figure(figsize=(10, 5))
plt.plot(distances)
plt.axhline(y=9, color='red', linestyle='--', label='Chosen eps = 9')
plt.title('K-Distance Graph (k=5) - used to choose DBSCAN eps')
plt.xlabel('Points sorted by distance')
plt.ylabel('5th Nearest Neighbor Distance')
plt.legend()
plt.grid()
plt.show()
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps=9, min_samples=5)
db_labels = dbscan.fit_predict(X2)

n_clusters = len(set(db_labels)) - (1 if -1 in db_labels else 0)
n_noise = list(db_labels).count(-1)
print(f"Clusters found        : {n_clusters}")
print(f"Outlier points flagged: {n_noise} out of {len(db_labels)}")

plt.figure(figsize=(10, 6))
mask = db_labels != -1
plt.scatter(X2[mask, 0], X2[mask, 1], c=db_labels[mask], cmap='rainbow', s=60, label='Clustered')
plt.scatter(X2[~mask, 0], X2[~mask, 1], c='black', marker='x', s=100, label='Outliers (noise)')
plt.title('DBSCAN Clustering - Income vs Spending Score')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.show()

if n_clusters > 1:
    print(f"Silhouette Score (DBSCAN, excluding noise): {silhouette_score(X2[mask], db_labels[mask]):.4f}")
```

## Summary: K-Means vs Hierarchical vs DBSCAN (Income vs Spending Score)

| Algorithm | Needs k upfront? | Handles outliers? | Cluster shape assumption |
|---|---|---|---|
| K-Means | Yes | No (forces every point into a cluster) | Spherical, similar-sized |
| Hierarchical (Agglomerative) | No (choose via dendrogram) | No | Flexible, but no explicit noise handling |
| DBSCAN | No (density-based) | Yes (explicit noise label) | Arbitrary shape, density-based |

K-Means and Hierarchical Clustering produce very similar 5-segment structures here (both silhouette scores land close together), which is a good sign that the segments are genuine rather than an artifact of one algorithm's assumptions. DBSCAN's value isn't a competing segmentation — it's the **outlier list**: those flagged points are the customers worth a second look before they're bucketed into any segment at all, which is exactly the kind of check a fintech risk/fraud team would run before applying segment-based offers or credit decisions.
